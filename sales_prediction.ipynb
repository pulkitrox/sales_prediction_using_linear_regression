Installing Libraries

```python
import numpy as np
import   matplotlib.pyplot as plt
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
```

Loading Dataset

```python
dataset = pd.read_csv('advertising_sales_data.csv')
df = pd.DataFrame(dataset)
```

Q1: Finding average expenditure of TV Advertising

```python
average_tv_advertising = df['TV'].mean()
print("Average TV Advertising Expenditure: ", average_tv_advertising)
```

Q2: Finding Correlation Index between Radio Advertising and Sales

```python
corr_radio_sales = df['Radio'].corr(df['Sales'])
print("Correlation Index between Radio Advertising and Sales: ", corr_radio_sales)
```

Q3: Finding Medium of Advertising with Highest Impact

```python
corr_matrix = df[['TV', 'Radio', 'Newspaper']].corrwith(df['Sales'])
highest_impact_medium = corr_matrix.idxmax()
highest_impact_value = corr_matrix.max()
print(f"Medium of Advertising with Highest Impact: {highest_impact_medium} --> {highest_impact_value} ")
```

Splitting Dataset into Independent Variable Matrix X and Dependent Variable Matrix y

```python
X = df.iloc[:, [1,2,3]].values
y = df.iloc[:, -1].values
print("X: ")
print(X)
print("y: ")
print(y)
```

Imputing X for Missing Values

```python
imputer = SimpleImputer()
imputer.fit(X)
X = imputer.transform(X)
print("Imputed X: ")
print(X)
```

Splitting Dataset into Training and Testing Sets, Performing Feature Scaling(Standardization) on X_train, X_test

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=7)
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
print("X_train: ")
print(X_train)
print("X_test: ")
print(X_test)
```

Creating Linear Regression Model for TV, Radio, Newspaper, using Standardization and Training it

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

Predictions on Test Set for Standardization Model

```python
y_pred = model.predict(X_test)
print("y_pred: ")
print(y_pred)
```

Q4: Visualizing Actual Sales vs Predicted Sales

```python
plt.scatter(y_test, y_pred, color='blue', alpha=0.6, label='Data Points')
min_val = min(y_test.min(), y_pred.min())
max_val = max(y_test.max(), y_pred.max())
plt.plot([min_val, max_val], [min_val, max_val], 'r--', lw=2, label='Perfect Fit')
m, b = np.polyfit(y_test, y_pred, 1)
plt.plot(y_test, m*y_test + b, color='green', linestyle='-.', label='Trend Line')
plt.xlabel('Actual Sales')
plt.ylabel('Predicted Sales')
plt.title('Actual vs. Predicted Sales')
plt.legend()
plt.show()
```

Q5: Predictions for TV,Radio,Newspaper : 200,40,50

```python
X_new = [[200,40,50]]
X_new = sc.transform(X_new)
y_new = model.predict(X_new)
print("Sales for New Data (200,40,50): ", y_new)
```

Evaluating Performance of TV, Radio, Newspaper using Standardization Model

```python
from sklearn.metrics import mean_squared_error

mse_all = mean_squared_error(y_test, y_pred)
rmse_all = np.sqrt(mse_all)
r2_all = r2_score(y_test, y_pred)

print("MSE for S:", mse_all)
print("RMSE for S:", rmse_all)
print("R² Score for S:", r2_all)
```
